# Importing the Required Libraries

```python
import pandas as pd
```

# Loading the Dataset

```python
dataset= pd.read_csv("expense.csv")
dataset.head()
```

# Filling The Null Values

```python
dataset['Amount'] = dataset['Amount'].fillna(0)
dataset.head()
```

# Creating the Series

```python
dataset['Amount']= pd.to_numeric(dataset['Amount'],errors='coerce')
dataset
```

```python
month_expense = pd.Series(data = dataset['Amount'].values, index = dataset['Month'])
sort_expense= month_expense.sort_index()
total_expense= sort_expense.sum()
print(f"Months :: {sort_expense}")
print(f"Total Expense :: {total_expense}")
```
