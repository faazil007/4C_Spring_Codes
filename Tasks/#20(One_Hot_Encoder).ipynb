# Importing Required Libraries

```python
from sklearn.preprocessing import OneHotEncoder
import pandas as pd
```

# Defining Custom Order

```python
temperature_order = ['Cold', 'Warm', 'Hot'] 
taste_order = ['Sour', 'Sweet', 'Salty'] 
```

# Sample dataset with Temperature,Taste and Target

```python
data = { 
    'Temperature': ['Hot', 'Cold', 'Warm', 'Warm', 'Hot', 'Cold'], 
    'Taste': ['Sour', 'Sweet', 'Salty', 'Sweet', 'Sour', 'Salty'] ,
    'Target': ['Good', 'Bad', 'Good', 'Bad', 'Good', 'Bad']
} 
```

# Converting Dictionary to Dataframe

```python
df = pd.DataFrame(data)
print(f" Now Dataframe is ::{df}")
```

# Intialize LabelEncoder with custom orders

```python
encoder = OneHotEncoder()
encoded_features = encoder.fit_transform(df[['Temperature', 'Taste']])
```

# Fitting

```python
encoded_df = pd.DataFrame(encoded_features, columns=encoder.get_feature_names_out(['Temperature', 'Taste'])
```

# Displaying the Dataframe

```python
print(" Dataframe is :: ")
print(encoded_df)
```
