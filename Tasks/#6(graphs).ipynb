# Grouped Bar Graph (1)

```python
#"In a school, the number of boys and girls participating in three sports is as follows:
# Cricket → Boys: 30, Girls: 20
# Football → Boys: 25, Girls: 15
# Tennis → Boys: 10, Girls: 12
# Draw a suitable graph."

import matplotlib.pyplot as plt
import numpy as np

sports = ['Cricket' , 'Football' , 'Tennis']
boys = [30,25,10]
girls=[20,15,12]
X = np.arange(len(sports))
bar_width = 0.4
plt.bar(X , boys , label = 'BOYS', width =0.4 , color = ['r' , 'r' , 'r'] , edgecolor = 'Black' , linewidth = 1.2)
plt.bar(X + bar_width ,girls, label = 'GIRLS',width =0.4 , color = ['b' , 'b' , 'b'] , edgecolor = 'Black' , linewidth = 1.2)
plt.title(" Sports Participation by Gender")
plt.xlabel("Sports")
plt.ylabel("Students")
plt.legend()
plt.xticks(X + bar_width / 2, sports)  # Centering the group
plt.show()
```

# Grouped Bar Graph (2)

```python
# The following table shows the marks obtained by two students, Ali and Sana, in five subjects:
# Math: Ali – 88, Sana – 92
# English: Ali – 76, Sana – 85
# Science: Ali – 90, Sana – 95
# Urdu: Ali – 70, Sana – 75
# Islamiat: Ali – 80, Sana – 83

# Draw a grouped bar graph to represent the comparison of marks.

subjects = ['Maths' , 'English' , 'Science' , 'Urdu' , 'Islamiyat']
Ali = [88 , 76 , 90 , 70 , 80]
Sana = [92,85,95,75,83]

X = np.arange(len(subjects))
bar_width = 0.4
plt.bar(X , Ali , label = 'ALI' , width = 0.4 , color =['b' , 'b' , 'b', 'b', 'b'] , edgecolor = 'Black', linewidth = 1.2 )
plt.bar(X + bar_width , Sana , label = 'SANA' , width = 0.4 , color =['g' , 'g' , 'g', 'g', 'g'] , edgecolor = 'Black', linewidth = 1.2 )
plt.title("Marks of Students")
plt.xlabel("Subjects")
plt.ylabel("Marks")
plt.legend()
plt.xticks(X + bar_width / 2 , subjects)
plt.show()
```

# Stacked Bar Graph 1

```python
# "A company sells three products (A, B, C) in 4 regions. The sales data (in units) is as follows:
# Region 1 → A: 30, B: 20, C: 25
# Region 2 → A: 35, B: 15, C: 30
# Region 3 → A: 25, B: 22, C: 18
# Region 4 → A: 40, B: 30, C: 10
# Draw a stacked bar chart to represent the sales data of the products in each region."


import matplotlib.pyplot as plt
import numpy as np

regions = ['Region 1', 'Region 2', 'Region 3', 'Region 4']
product_A = [30, 35, 25, 40]
product_B = [20, 15, 22, 30]
product_C = [25, 30, 18, 10]

x = np.arange(len(regions))

plt.bar(x, product_A, label='Product A', color='#66b3ff')
plt.bar(x, product_B, bottom=product_A, label='Product B', color='#99ff99')
plt.bar(x, product_C, bottom=np.array(product_A)+np.array(product_B), label='Product C', color='#ff9999')

plt.xticks(x, regions)
plt.xlabel('Regions')
plt.ylabel('Units Sold')
plt.title('Stacked Bar Chart - Product Sales by Region')
plt.legend()
plt.show()
```

# PIE Chart 2

```python
# A survey was conducted to find out the preferred mode of transport among 100 people. The results are:

# Car: 40 people

# Bus: 25 people

# Bike: 20 people

# Walk: 15 people

# Draw a pie chart to represent the data.

import matplotlib.pyplot as plt

# Data
transport_modes = ['Car', 'Bus', 'Bike', 'Walk']
people = [40, 25, 20, 15]

# Create pie chart
plt.pie(people, labels=transport_modes, autopct='%1.1f%%', startangle=45)

# Title
plt.title("Preferred Mode of Transport (Survey of 100 People)")

# Equal aspect ratio ensures the pie is a circle
plt.axis('equal')

# Show the chart
plt.show()
```

# Nested PIE Chart 1

```python
# A tech company has 2 departments:
# Development (60 employees) and Marketing (40 employees).

# In Development, there are 3 roles:

# Frontend: 25 employees

# Backend: 20 employees

# QA: 15 employees

# In Marketing, there are 2 roles:

# Content: 25 employees

# SEO: 15 employees

# Draw a nested pie chart to show distribution of employees across departments and their roles.

import matplotlib.pyplot as plt

# Outer layer (Departments)
outer_labels = ['Development', 'Marketing']
outer_sizes = [60, 40]
outer_colors = ['#66b3ff', '#ff9999']

# Inner layer (Roles)
inner_labels = ['Frontend', 'Backend', 'QA', 'Content', 'SEO']
inner_sizes = [25, 20, 15, 25, 15]
inner_colors = ['#99ff99', '#ffcc99', '#ffb3e6', '#c2c2f0', '#ff6666']

# Create figure and axis
fig, ax = plt.subplots(figsize=(8, 8))  # Increased figure size for clarity
fig.suptitle("Company Employee Distribution (Nested Pie Chart)", fontsize=14, fontweight='bold', y=0.95)

# Outer Pie
ax.pie(outer_sizes, labels=outer_labels,labeldistance=1.02,autopct='%1.1f%%', pctdistance=0.9, radius=1.3, colors=outer_colors,
       wedgeprops=dict(width=0.3, edgecolor='black', linewidth = 1.2))

# Inner Pie
ax.pie(inner_sizes, labels=inner_labels,labeldistance=0.5,autopct='%1.1f%%',  pctdistance=0.8,radius=1.0, colors=inner_colors,
       wedgeprops=dict(width=0.3, edgecolor='black' , linewidth = 1.2))

# Title
#plt.title("Company Employee Distribution (Nested Pie Chart)")

# Display
plt.show()

```

# Nested PIE Chart 2

```python
# "In a school, there are two main streams: Science and Arts. 
# Science has 50 students in Physics, 30 in Chemistry, and 20 in Biology.
# Arts has 40 in History and 30 in Literature.
# Draw a nested pie chart showing overall stream-wise and subject-wise distribution of students."

outer_labels = ['Science' , 'Arts']
outer_values = [100,70]
outer_colors = ['#66b3ff', '#ff9999']


inner_labels=['Physics' , 'Chemistry' , 'Biology']
inner_values = [50 , 30 , 20]
inner_colors =  ['#99ff99', '#ffcc99', '#ffb3e6']

fig,ax = plt.subplots(figsize=(8,8))
fig.suptitle(" Education Department" , fontsize = 14 , fontweight = 'bold' , y= 0.95)

ax.pie(outer_values , labels = outer_labels , autopct='%1.1f%%',labeldistance= 0.9  , pctdistance= 0.95 , radius = 1.3 , colors= outer_colors,
       wedgeprops= dict(width = 0.5 , edgecolor  = 'black' , linewidth = 1.2)
        )

ax.pie(inner_values , labels = inner_labels , autopct = '%1.1f%%', labeldistance= 0.9 , radius = 1.0 , pctdistance=0.97 , colors = inner_colors,
       wedgeprops=dict(width = 0.5 , edgecolor = 'black' , linewidth=1.2))

plt.show()

```

# Histogram 1

```python
# "The math test scores of 20 students are recorded as follows:
# 35, 45, 50, 55, 60, 65, 66, 68, 70, 72, 73, 74, 75, 78, 80, 82, 85, 90, 92, 95
# Draw a histogram to show the distribution of scores."


import matplotlib.pyplot as plt

# Test scores of 20 students
scores = [35, 45, 50, 55, 60, 65, 66, 68, 70, 72, 73, 74, 75, 78, 80, 82, 85, 90, 92, 95]

# Create histogram with bins (ranges)
plt.hist(scores, bins=[30,40, 50, 60, 70, 80, 90, 100], color='#66b3ff', edgecolor='black', linewidth=1.5)

plt.xlabel('Score Range')
plt.ylabel('Number of Students')
plt.title('Histogram of Math Test Scores')
plt.show()
```
