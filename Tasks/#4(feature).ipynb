# Importing the required libraries

```python
import numpy as np
```

# Creation of Array

```python
exam_scores = np.array([65,72,80,85,90,95,100])
min_score = np.min(exam_scores)
max_score = np.max(exam_scores)
```

# Range Scores

```python
range_scores = max_score - min_score
```

# Apply Min-Max Scaling

```python
min_max_scaled = (exam_scores - min_score) / range_scores
```

# Output the Values

```python
print("Original Exam Scores:", exam_scores) 
print("Min-Max Scaled Scores:", min_max_scaled) 
```
