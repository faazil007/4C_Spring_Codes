# Importing Required Libraries

```python
import numpy as np
import pandas as pd
from collections import Counter
from imblearn.under_sampling import NearMiss
```

# Version 1

```python
X = np.array([[1,1],[9,1],
              [2,1],[3,6],
              [4,5],[6,9]

              ])
Y= np.array([1,1,0,0,0,0])
print(Counter(Y))
data = pd.DataFrame(X)
print(data)
nm = NearMiss(n_neighbors=2,version = 1)
X_res,Y_res=nm.fit_resample(X,Y)
print(X_res)
print(Y_res)

```
