# Importing required Libraries

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder
```

# Loading the dataset

```python
data = pd.read_csv("../Datasets/random_dataset_with_missing_values.csv")
data
```

# Mean Technique

```python
df_mean_filled = data.copy()

# Numerical columns ke liye mean fill
df_mean_filled['Age'] = df_mean_filled['Age'].fillna(df_mean_filled['Age'].mean())
df_mean_filled['Salary'] = df_mean_filled['Salary'].fillna(df_mean_filled['Salary'].mean())
#df_mean_filled['Department'] = df_mean_filled['Department'].fillna(df_mean_filled['Department'].mean())

df_mean_filled.head(10)
```

# Median Technique

```python
df_median_filled = data.copy()

# Numerical columns ke liye median fill
df_median_filled['Age'] = df_median_filled['Age'].fillna(df_median_filled['Age'].median())
df_median_filled['Salary'] = df_median_filled['Salary'].fillna(df_median_filled['Salary'].median())

df_median_filled.head(10)
```

# Mode Technique

```python
df_mode_filled = data.copy()

# Har column ke liye mode se fill
for column in ['Age', 'Salary', 'Department']:
    mode_value = df_mode_filled[column].mode()[0]
    df_mode_filled[column] = df_mode_filled[column].fillna(mode_value)

df_mode_filled.head(10)
```

# Arbitrary Technique

```python
df_arbitrary_filled = data.copy()

# Custom arbitrary values se fill karna
df_arbitrary_filled['Age'] = df_arbitrary_filled['Age'].fillna(99)
df_arbitrary_filled['Salary'] = df_arbitrary_filled['Salary'].fillna(99999)
df_arbitrary_filled['Department'] = df_arbitrary_filled['Department'].fillna('Unknown')

df_arbitrary_filled.head(10)
```

# End of distribution

```python
df_end_dist_filled = data.copy()

# End of distribution: mean + 3*std
for column in ['Age', 'Salary']:
    end_value = df_end_dist_filled[column].mean() + 3 * df_end_dist_filled[column].std()
    df_end_dist_filled[column] = df_end_dist_filled[column].fillna(end_value)

# For categorical column, we can treat end distribution as "Rare" value
df_end_dist_filled['Department'] = df_end_dist_filled['Department'].fillna('Rare')

df_end_dist_filled.head(10)
```
