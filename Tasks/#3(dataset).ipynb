# Importing Required Libraries

```python
import pandas as pd
import matplotlib.pyplot as plt
```

# Loading the Data Set


```python
dataset= pd.read_csv("studentrecord.csv")
dataset.head()
```

# Showing the Column Names and Data Types

```python
print(dataset.dtypes)
```

# To Display Summary Statistics of Numerical Columns

```python
print(dataset.describe())
```

# Average Score in Science

```python
average_science = dataset['Science'].mean()

# Display result
print("Average Science Score:", average_science)
```

# Adding a New Column

```python
dataset['Total'] = dataset['Math'] + dataset['Science'] + dataset['English']

# Display the updated DataFrame
print(dataset)
```

# Adding Another Column

```python
dataset['Result'] = dataset['Total'].apply(lambda x: 'Pass' if x >= 240 else 'Fail')

# Display the updated DataFrame
print(dataset)

# Count how many passed and failed
result_counts = dataset['Result'].value_counts()

print("\nSummary:")
print("Passed Students:", result_counts.get('Pass', 0))
print("Failed Students:", result_counts.get('Fail', 0))
```

# Filter only female students who scored more than 80 in math

```python
filtered_df = dataset[(dataset['Gender'] == 'Female') & (dataset['Math'] > 80)]

# Display the filtered result
print(filtered_df)
```

# Plotting a Bar Chart

```python
plt.figure(figsize=(10,6))
plt.bar(dataset['Name'], dataset['Total'], color='skyblue')
plt.title('Total Scores by Student')
plt.xlabel('Student Name')
plt.ylabel('Total Score')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

# Create a Pie Chart

```python
plt.figure(figsize=(6,6))
plt.pie(result_counts, labels=result_counts.index, autopct='%1.1f%%', startangle=140, colors=['green', 'red'])
plt.title('Distribution of Pass/Fail Results')
plt.axis('equal')  # Make the pie chart a circle
plt.show()
```

# Plot a Histogram

```python
plt.figure(figsize=(8,5))
plt.hist(dataset['Math'], bins=5, color='skyblue', edgecolor='black')
plt.title('Distribution of Math Scores')
plt.xlabel('Math Score')
plt.ylabel('Number of Students')
plt.grid(True)
plt.tight_layout()
plt.show()
```
