# Dictionary

```python
students = {
    "S101": {
        "Name": "Ali", "Age": 19, "Subjects": ["DL", "NLP"], "Grades": {"DL": 64, "NLP": 86}
    },
    "S102": {
        "Name": "Saim", "Age": 20, "Subjects": ["AI", "ML"], "Grades": {"AI": 64, "ML": 86}
    }
}
```

# Adding a New Record

```python
students["S103"] = {
    "Name": "Kamran", "Age": 20, "Subjects": ["DL", "PAI"], "Grades": {"PAI": 88, "DL": 91}
}
```

# Updating Age of Kamran from 20 to 21

```python
students["S103"].update({"Age": 21})

print(students)
```

# Removing Ali's NLP

```python
del students["S101"]["Grades"]["NLP"]
```

# Showing all Students With PAI

```python
for s in students.values():
    if "PAI" in s["Subjects"]:
        print(s["Name"])
```
