```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import LabelEncoder, StandardScaler

# Load Data
data = pd.read_csv("spotify-2023.csv", encoding='latin-1')

# Separate 'streams' first
target = data['streams']

# Encode only categorical (object) columns
encoder = LabelEncoder()
for i in data.columns:
    data[i] = encoder.fit_transform(data[i])
# Put the original target back
data['streams'] = target

# Feature Scaling (excluding 'streams')
features = data.drop(columns=['streams'])
scaler = StandardScaler()
scaled_features = scaler.fit_transform(features)
scaled_df = pd.DataFrame(scaled_features, columns=features.columns)
scaled_df['streams'] = data['streams']

# No feature drop here
X = scaled_df.drop(columns=['streams'])
Y = scaled_df['streams']

# Train-test split
x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size=0.2)

# Model Training
model = LinearRegression()
model.fit(x_train, y_train)

# Prediction & Evaluation
y_pred = model.predict(x_test)
RMSE = mean_squared_error(y_test, y_pred)

print(f"Root Mean Squared Error: {RMSE:.2f}")
plt.scatter(y_pred, y_test)
plt.xlabel("Predicted Streams")
plt.ylabel("Actual Streams")
plt.title("Prediction vs Actual")
plt.show()

```
