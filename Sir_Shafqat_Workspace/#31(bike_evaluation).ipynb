# Importing required Libraries

```python
import pandas as pd
import seaborn as sns
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,mean_absolute_error
from sklearn.preprocessing import LabelEncoder,StandardScaler
from sklearn.tree import DecisionTreeRegressor,DecisionTreeClassifier
import warnings
warnings.filterwarnings("ignore")
from collections import Counter
```

# Loading the dataset

```python
data = pd.read_csv("../Datasets/bike_sales_100k.csv")
data.head()
```

# Encoding the labels

```python
encoder = LabelEncoder()
for i in data.columns:
    data[i] = encoder.fit_transform(data[i])

data    
```

# Scaling

```python

scaler = StandardScaler()
data['Price'] = scaler.fit_transform(data[['Price']])
```

# Separating X & Y 

```python
X = data.drop(columns=['Customer_Gender','Sale_ID','Customer_ID','Date','Salesperson_ID'])
Y = data['Customer_Gender']
```

# Splitting into training and testing

```python
X_train,X_test,y_train,y_test = train_test_split(X,Y,test_size = 0.2,random_state =10)
```

# Loading the model

```python
model = DecisionTreeClassifier(max_depth=10)
model.fit(X_train,y_train)
```

# Predictions and Error

```python
predictions = model.predict(X_test)
#MAR = mean_absolute_error(y_test,predictions)
accuracy = accuracy_score(y_test,predictions)
#print(f" Error is :: {MAR}")
print(f" Accuracy is :: {accuracy}")
```
