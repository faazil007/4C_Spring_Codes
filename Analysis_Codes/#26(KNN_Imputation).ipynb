# Importing required LIbraries

```python
from sklearn.impute import SimpleImputer
import pandas as pd
import numpy as np
from sklearn.impute import KNNImputer
```

# Imputing Technique Using Mean

```python
imputer = SimpleImputer(strategy = 'mean')
data = pd.DataFrame({
    'C1': [1,3,2,12,np.nan,50]

})
imputerr = imputer.fit_transform(data)
print(type(imputerr))

imp_data = pd.DataFrame(imputerr)
print(imp_data)
print(type(imp_data))
```

# Imputation Using KNN Imputator

## For Signal Column

```python
data = pd.DataFrame({
    'Age': [21,25,np.nan,27,28]

})
KNN = KNNImputer(n_neighbors = 2)
imputerr =KNN.fit_transform(data)
imputed_data = pd.DataFrame(imputerr)
print(imputed_data)






```

## For Multiple Columns

```python
data = pd.DataFrame({
    'ID': [1,2,3,4],
    'F1': [2.1,np.nan,2.5,2.7],
    'F2': [3.1,np.nan,3.3,3.4],
    'F3': [1.5,1.3,np.nan,1.6]
})
print(data) 
KNN = KNNImputer(n_neighbors = 2)
imputerr =KNN.fit_transform(data)
imputed_data = pd.DataFrame(imputerr)
print(imputed_data)


```
